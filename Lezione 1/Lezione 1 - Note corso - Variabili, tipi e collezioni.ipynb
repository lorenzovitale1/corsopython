### E14 · Classificare la presenza

**Perché ci serve:** Dare uno stato comprensibile a ogni presenza e separare i dati non validi dalle presenze realmente registrate.

Scrivi una catena if/elif/else: ore fuori da 0–8 → Dato non valido; 8 → Presenza completa; 6 o 7 → Soglia raggiunta; 0–5 → Soglia non raggiunta. Prova in ordine 5, 6, 8, 9 e -1. Il controllo di validità viene prima degli altri.

**Controllo del risultato:** 5 → Soglia non raggiunta; 6 → Soglia raggiunta; 8 → Presenza completa; 9 e -1 → Dato non valido.


**Dati da usare nella tua soluzione:**



#### Esempio svolto con altri dati

L’ordine dei rami conta. Prima escludi i dati non validi, poi controlla il caso più specifico.


````python
ore = 7
if ore < 0 or ore > 8:
    print("Dato non valido")
elif ore == 8:
    print("Presenza completa")
elif ore >= 6:
    print("Soglia raggiunta")
else:
    print("Soglia non raggiunta")
````

**Perché ci serve:** Distinguere un inserimento utilizzabile da uno da correggere, mostrando un messaggio comprensibile all’operatore.

Chiedi un numero di ore. Prova a convertirlo in int con try/except ValueError. Se si converte, controlla con if che sia tra 0 e 8 inclusi. Nel finally stampa "Controllo terminato". Prova 6, otto, -1 e 9.

**Controllo del risultato:** 6 → Dato valido; otto → Numero intero richiesto; -1 e 9 → Fuori intervallo. In tutti i casi compare Controllo terminato.


#### Esempio svolto con altri dati

La conversione può fallire; un intero convertito può comunque essere fuori intervallo. finally esegue il messaggio conclusivo in entrambi i casi.


````python
testo = "sette"
try:
    ore = int(testo)
    if 0 <= ore <= 8:
        print("Dato valido")
    else:
        print("Fuori intervallo")
except ValueError:
    print("Numero intero richiesto")
finally:
    print("Controllo terminato")
````

**Perché ci serve:** Conservare le presenze utilizzabili e quantificare gli scarti senza bloccare l’intero lavoro per un solo dato errato.

Visita i testi con un for. Raccogli in una lista solo gli interi tra 0 e 8. Conta gli altri come scarti. Gestisci ValueError senza interrompere il ciclo. Prova anche lista vuota e ["x", "-1"].

**Controllo del risultato:** [8, 6] e 3 scarti; lista vuota → [] e 0; ["x", "-1"] → [] e 2.


**Dati da usare nella tua soluzione:**

```python
testi = ["8", "x", "6", "-1", "9"]
```


#### Esempio svolto con altri dati

try sta dentro il ciclo: un testo sbagliato non interrompe la lettura dei successivi. Ogni scarto aumenta il contatore una volta.


````python
testi = ["7", "errore", "0", "10"]
validi = []
scarti = 0
for testo in testi:
    try:
        ore = int(testo)
    except ValueError:
        scarti = scarti + 1
        continue
    if 0 <= ore <= 8:
        validi.append(ore)
    else:
        scarti = scarti + 1
print(validi, scarti)
````
