| Tipo | Come si scrive | Ordine e duplicati | Come leggo un dato | Cosa posso cambiare | Quando la uso | Esempi |
|---|---|---|---|---|---|---|
| **Lista** · `list` | `["A", "B", "A"]` · vuota `[]` | Mantiene l’ordine. Ammette ripetizioni. | `letture[0]` per la prima, `letture[-1]` per l’ultima | `letture.append("C")` aggiunge in fondo. `letture[0] = "X"` sostituisce. | Quando mi serve una sequenza ordinata che posso aggiornare e che ammette ripetizioni. | Lista della spesa da aggiornare. Cronologia degli accessi che registra anche i passaggi ripetuti. |
| **Tupla** · `tuple` | `("B2", "mattina")` · vuota `()` · singola `("B2",)` | Mantiene l’ordine. Ammette ripetizioni. | `sede[0]` oppure `aula, turno = sede` | Le posizioni sono fisse. Per cambiare aula creo una nuova tupla. | Quando voglio tenere insieme valori con un significato preciso per posizione e mantenere fissi gli elementi della sequenza. | Coordinate `(10, 20)` con `x` e `y`. Risoluzione `(1920, 1080)` con larghezza e altezza. |
| **Insieme** · `set` | `{"A", "B"}` · vuoto `set()` | Ogni elemento compare una volta. L’ordine di visita non è garantito. | `"A" in presenti` controlla l’appartenenza. Nessun indice. | `presenti.add("C")` aggiunge. `presenti.discard("A")` rimuove se presente. | Quando mi interessano i valori distinti, voglio verificare se un elemento è presente o confrontare due gruppi. | Partecipanti contati una sola volta. Città visitate da entrambe le persone. |
| **Dizionario** · `dict` | `{"aula": "B2", "posti": 8}` · vuoto `{}` | Mantiene l’ordine di inserimento. Chiavi uniche, valori anche ripetuti. | `scheda["posti"]` legge il valore. `"posti" in scheda` cerca la chiave. | `scheda["posti"] = 10` aggiorna il campo. Una chiave nuova aggiunge un campo. | Quando voglio descrivere una scheda con campi nominati o trovare un valore a partire da una chiave. | Prezzo per codice prodotto. Numero di telefono per nome del contatto. |

**Altre scritture da riconoscere.** `list("AB")` produce una lista di caratteri. `tuple(letture)` conserva ordine e ripetizioni. `set(letture)` conserva soltanto i valori distinti. `dict([("A", 2), ("B", 5)])` costruisce un dizionario da coppie chiave-valore. `len(...)` conta gli elementi di ciascuna collezione, quindi in un dizionario conta le chiavi.

**Due dettagli di sintassi.** In `("B2",)` è la virgola a creare la tupla di un elemento. Le graffe vuote `{}` creano un dizionario. Come elementi di un set e come chiavi di un dizionario puoi usare, per esempio, stringhe, numeri e tuple di stringhe. Liste e dizionari non possono svolgere questi due ruoli.

## Quanto affrontato finora

```python
# 1. VARIABILI E TIPI
nome = "Ada"                # str   → testo
ore = 6                     # int   → numero intero
quota = 25.50               # float → numero decimale (in Python si usa il punto)
confermato = True           # bool  → True oppure False

print(nome, ore)            # Mostra più valori (la virgola li separa con uno spazio nella stampa)
print(type(ore))            # Mostra il tipo di una variabile


# 2. INPUT E CONVERSIONI
nome = input("Come ti chiami? ")   # input() restituisce sempre una stringa
ore = int("6")                      # Dal testo "6" all'intero 6
quota = float("25.50")              # Dal testo "25.50" al float 25.50
testo = str(ore)                    # Dall'intero 6 al testo "6"


# 3. CALCOLI E CONFRONTI
print(ore + 2, ore - 1, ore * 2)    # Somma, sottrazione, moltiplicazione
print(ore / 8 * 100)                # Percentuale di presenza: 75.0
print(95 // 60, 95 % 60)            # 95 minuti → 1 ora intera e 35 minuti

# = assegna un valore; == confronta due valori
print(ore == 6, ore != 0)           # Uguale, diverso
print(ore >= 6, ore < 8)            # Ogni confronto restituisce True o False
print(ore >= 6 and confermato)      # True se entrambe le condizioni sono vere


# 4. STRINGHE
nome = "  aDA roSSi  ".strip().title()  # strip() toglie gli spazi esterni, title() mette in maiuscolo la prima lettera di ogni parola e in minuscolo le altre → "Ada Rossi"
codice = "PY-042"
print(codice[0])                    # Primo carattere: "P" (si parte da indice 0)
print(codice[-1])                   # Ultimo carattere: "2"
print(codice[:2], codice[-3:])      # Porzioni di testo: "PY" e "042"
print(nome.upper(), nome.lower())   # Tutto maiuscolo / tutto minuscolo
print(f"{nome}: {ore} ore, quota {quota:.2f} euro")  # Importo con due decimali


# 5. LISTA [] → valori in ordine, modificabili, anche ripetuti
iscritti = ["Ada", "Leo", "Mia"]
iscritti.append("Eva")              # Aggiunge un elemento in fondo
print(iscritti[0], iscritti[-1])    # Primo e ultimo elemento
print(iscritti[0:2])                # Elementi agli indici 0 e 1 (il 2 è escluso)
print(len(iscritti))                # Numero di elementi: 4
print("Leo" in iscritti)            # Controlla se il nome è presente

copia = iscritti.copy()             # Crea una copia della lista
copia[0] = "Anna"                   # In iscritti il primo nome resta "Ada"


# 6. TUPLA () → valori in ordine, con posizioni fisse
lezione = ("B2", "mattina")
aula, turno = lezione               # Spacchettamento: un valore a ciascun nome
print(aula, turno)                  # B2 mattina
aula_singola = ("B2",)              # Per una tupla di un elemento serve la virgola


# 7. DIZIONARIO {} → associa ogni chiave a un valore
scheda = {"nome": "Ada", "ore": 6}
print(scheda["nome"])               # Legge il valore tramite la chiave
scheda["ore"] = 7                   # Aggiorna un campo esistente
scheda["aula"] = "B2"               # Aggiunge un nuovo campo
print(scheda.get("email", "assente"))  # Valore alternativo se manca la chiave
print(list(scheda.items()))         # Mostra le coppie (chiave, valore)


# 8. SET → valori distinti, senza posizioni indicizzate
letture_badge = ["Ada", "Leo", "Ada"]
presenti = set(letture_badge)        # Ada compare una sola volta nel set
presenti.add("Mia")                 # Aggiunge Mia ai presenti
assenti = set(iscritti) - presenti  # Differenza: iscritti che mancano
print(len(letture_badge))           # 3 passaggi registrati dal lettore
print(sorted(assenti))              # ['Eva']; sorted() ordina i nomi per la stampa

# Collezioni vuote: [] lista, () tupla, {} dizionario, set() insieme
```

```python
def potenza(base, esp=2):
    return base ** esp

print(potenza(5))  # 25
print(potenza(2, 10))  # 1024

def aggiungi(x, lista=None):
    if lista is None:
        lista = []
    lista.append(x)
    return lista
```

Funzioni 6

Dopo aver eseguito la cella, prova `somma_tutti(*[2, 3])`. L’asterisco nella chiamata passa i due numeri separatamente.



```python
def somma_tutti(*numeri):
    return sum(numeri)

print(somma_tutti(1, 2, 3, 4))  # 10

def scheda(**dati):
    for k, v in dati.items():
        print(k, v)

scheda(nome="Ada", eta=36)
```

### R17 Il ciclo for

Fondamenti 48

Un `for` può contenere un `if` per selezionare solo alcuni elementi.



```python
voti = [28, 30, 25]
for v in voti:
    if v >= 28:
        print(v, "ottimo")
    else:
        print(v, "buono")
```

### R19 Leggere e aggiornare un dizionario

Fondamenti 39

`d.items()` fornisce le coppie chiave-valore. `d.get(chiave, 0)` restituisce zero se la chiave manca.



```python
d = {"a": 1, "b": 2}
d["c"] = 3
del d["a"]
print("b" in d)  # True
print(list(d.keys()))  # ["b", "c"]
print(list(d.items()))
```
