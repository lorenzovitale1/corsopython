````python
from pathlib import Path
from uuid import uuid4
from zipfile import ZipFile, ZipInfo, ZIP_DEFLATED
import json

cartella = Path.cwd()
nome_notebook = "Lezione 4 - E01 - Ripasso.ipynb"
with (cartella / nome_notebook).open(encoding="utf-8") as file:
    salvato = json.load(file)

# Nella copia da consegnare teniamo il testo delle celle e le lettere salvate.
celle = []
for cella in salvato["cells"]:
    copia = {k: cella[k] for k in ("cell_type", "id", "source")}
    copia["metadata"] = {}
    if copia["cell_type"] == "code":
        copia.update(execution_count=None, outputs=[])
    celle.append(copia)
pulito = {"nbformat": 4, "nbformat_minor": 5, "metadata": {}, "cells": celle}

contenuti = {nome_notebook: json.dumps(pulito, ensure_ascii=False).encode("utf-8")}
for nome in ("rubrica.json", "contatto.py", "main.py", "riepilogo.csv"):
    if (cartella / nome).is_file():
        contenuti[nome] = (cartella / nome).read_bytes()

nome_consegna = "ripasso_" + uuid4().hex[:12]
pronta = cartella / nome_consegna
pronta.mkdir()
for nome, dati in contenuti.items():
    (pronta / nome).write_bytes(dati)
nome_zip = nome_consegna + ".zip"
with ZipFile(cartella / nome_zip, "x") as archivio:
    for nome, dati in contenuti.items():
        archivio.writestr(ZipInfo(nome_consegna + "/" + nome), dati, compress_type=ZIP_DEFLATED)
print("Cartella da caricare:", nome_consegna)
print("Alternativa ZIP:", nome_zip)
print("File inclusi:", ", ".join(contenuti))
````
